```python
import pandas as pd
import numpy as np

import sklearn 
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler,LabelEncoder
import pickle
```

```python
#first step load your file
data=pd.read_csv('/Users/riteshchaudhary/Downloads/CONSULTANCY/DEEP_Learning_project_01/Churn_Modelling.csv')
data.head(5)
```

```python
#feature droping delete unessential columns
data=data.drop(['RowNumber'	,'CustomerId','Surname'],axis=1)
data.head(5)
```

```python
#ENCODE CATEGORICAL COLUMNS Eg: geography and gender
lable_encoder_gender=LabelEncoder()
data['Gender']=lable_encoder_gender.fit_transform(data['Gender'])
data.head(4)
```

```python
#ENCODE GEOGRAPY INTO NUMERIC BY USING ONEHOT ENCODER IT CONTERT TEXT VALUES BETWEEN 0 TO 1 NOT ONLY O OR 1
'''from sklearn.preprocessing import OneHotEncoder
onehotencode_geograpy=OneHotEncoder()
data['Geography']=onehotencode_geograpy.fit_transform(data['Geography'])
data.head(100)
#data['Gender'].values'''
```

```python
from sklearn.preprocessing import OneHotEncoder
import pandas as pd

# Create encoder
onehot = OneHotEncoder()

# Fit & transform
geo_encoded = onehot.fit_transform(data[['Geography']]).toarray()

# Convert into DataFrame
geo_encoded_df = pd.DataFrame(
    geo_encoded,
    columns=onehot.get_feature_names_out(['Geography'])
)

# Drop old column and join new columns
data = pd.concat([data.drop('Geography', axis=1), geo_encoded_df], axis=1)

data.head()
```

```python
import pickle

# Save LabelEncoder
with open('label_encoder_gender.pkl', 'wb') as f:
    pickle.dump(lable_encoder_gender, f)

# Save OneHotEncoder
with open('onehot_encoder_geo.pkl', 'wb') as f:
    pickle.dump(onehot, f)
```

```python
# DEVIDE DATA INTO INDEPENDENT AND DEPENDENT FEATURE
## Indpendent and dependent features

from sklearn.model_selection import train_test_split# 2d dataframe
X = data.drop(['Exited'], axis=1)  # 1 d dataframe
y = data['Exited']
```

```python
#TRAIN TEST SPLIT
X_train,X_test,y_train,y_test=train_test_split(X,y,random_state=42,test_size=.20)

#SCALE THESE FEATURES
scaler=StandardScaler()
X_train=scaler.fit_transform(X_train)
X_test=scaler.transform(X_test)

#SAVE PICKLE FILE FRO SCALE DATA
with open('scaler.pkl', 'wb') as f:
    pickle.dump(scaler, f)
```

ANN MODEL

```python

import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.callbacks import EarlyStopping, TensorBoard
import datetime
''' tensorflow as tf → main deep learning library
Sequential → model type (layer-by-layer network)
Dense → fully connected layer (ANN neuron layer)
EarlyStopping → stops training when no improvement
TensorBoard → visualize training (graphs, loss)
datetime → used to create log folder name'''
```

    BUILD MODEL

```python
(X_train.shape[1],)
```

```python
model = Sequential([#Create ANN model (layer-by-layer)
    Dense(64, activation='relu', input_shape=(X_train.shape[1],)),#FIRST HIDDEN  LAYER #64=64 NUERON
    Dense(32, activation='relu'),#SECOND HIDDEN LAYER
    Dense(1, activation='sigmoid')#OUTPUT LAYER #1 neuron → binary classification
                                                #sigmoid → output between 0–1
])
```

```python
model.summary()
```

```python
#OPTIMISER +LOSS
import tensorflow

opt = tensorflow.keras.optimizers.Adam(learning_rate=0.01)
loss = tensorflow.keras.losses.BinaryCrossentropy()
```

```python
#COMPILE THE MODEL
model.compile(optimizer=opt, loss='binary_crossentropy', metrics=['accuracy'])
```

```python
#TENSOR FLOW SETUP
from tensorflow.keras.callbacks import TensorBoard
import datetime

log_dir = "logs/fit/" + datetime.datetime.now().strftime("%Y%m%d-%H%M%S")

tensorboard_callback = TensorBoard(log_dir=log_dir, histogram_freq=1)
```

```python
#EARLY STOPPING
from tensorflow.keras.callbacks import EarlyStopping

early_stopping_callback = EarlyStopping(
    monitor='val_loss',
    patience=5,
    restore_best_weights=True
)
```

```python
# TRAIN THE MODEL
history = model.fit(
    X_train, y_train,
    validation_data=(X_test, y_test),
    epochs=100,
    callbacks=[tensorboard_callback, early_stopping_callback]
)
```

```python
#SAVE MODEL
model.save('model.h5')
```
